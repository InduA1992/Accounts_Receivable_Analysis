## Feature Engineering

## Create payment status

```python
df_clean["payment_status"] = np.where(
    df_clean["days_overdue_delay"] < 0,
    "Early Payment",
    np.where(
        df_clean["days_overdue_delay"] == 0,
        "On-Time Payment",
        "Late Payment"
    )
)

df_clean["payment_status"].value_counts()
```

## Calculate payment cycle

```python
df_clean["payment_cycle_days"] = (
    df_clean["clearing_date"] -
    df_clean["doc_date"]
).dt.days

df_clean["payment_cycle_days"].describe()
```

##  Extract year, month and quarter

```python
df_clean["clearing_year"] = df_clean["clearing_date"].dt.year
df_clean["clearing_month"] = df_clean["clearing_date"].dt.month_name()
df_clean["clearing_quarter"] = df_clean["clearing_date"].dt.quarter
df_clean[[
"clearing_date",
"clearing_year",
"clearing_month",
"clearing_quarter"
]].head()
```

## Customer payment analysis

```python
customer_metrics = (
    df_clean.groupby("cust_num")
    .agg(
        total_invoices=("document_no", "count"),
        total_invoice_amount=("amount", "sum"),
        average_delay=("days_overdue_delay", "mean"),
        maximum_delay=("days_overdue_delay", "max"),
        early_payments=(
            "payment_status",
            lambda x: (x == "Early Payment").sum()
        ),
        on_time_payments=(
            "payment_status",
            lambda x: (x == "On-Time Payment").sum()
        ),
        late_payments=(
            "payment_status",
            lambda x: (x == "Late Payment").sum()
        )
    )
    .reset_index()
)

customer_metrics["late_payment_rate"] = (
    customer_metrics["late_payments"] /
    customer_metrics["total_invoices"] * 100
)

customer_metrics["average_delay"] = (
    customer_metrics["average_delay"].round(2)
)

customer_metrics["late_payment_rate"] = (
    customer_metrics["late_payment_rate"].round(2)
)

display(customer_metrics.head())
```
